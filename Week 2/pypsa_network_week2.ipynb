```python
import pypsa
import pandas as pd

n = pypsa.Network()
```

```python
n.add("Bus", "bus_1", v_nom=11)
n.add("Bus", "bus_2", v_nom=11)
n.add("Bus", "bus_3", v_nom=11)
```

```python
n.add("Load", "load_1", bus="bus_1", p_set=500)

n.add(
    "Generator",
    "coal",
    bus="bus_1",
    p_nom=300,
    p_min_pu=50/300,
    marginal_cost=3,
)
```

```python
n.add(
    "Generator",
    "gas",
    bus="bus_2",
    p_nom=400,
    p_min_pu=100/400,
    marginal_cost=4,
)
```

```python
n.add("Load", "load_3", bus="bus_3", p_set=0)

n.add(
    "Generator",
    "solar",
    bus="bus_3",
    p_nom=300,
    marginal_cost=0,
)
```

```python
n.set_snapshots(pd.date_range("2025-01-01", periods=24, freq="h"))

n.generators_t.p_max_pu["solar"] = [
    0, 0, 0, 0, 0, 0,
    0.1, 0.3, 0.5, 0.7, 0.9, 1.0,
    1.0, 0.9, 0.7, 0.5, 0.3, 0.1,
    0, 0, 0, 0, 0, 0
]

n.loads_t.p_set["load_3"] = [100, 100, 100, 100, 120, 120,
                             150, 180, 200, 200, 200, 200,
                             180, 180, 180, 200, 200, 200,
                             200, 200, 200, 200, 150, 120]
```

```python
n.add(
    "Line",
    "line_1",
    bus0="bus_1",
    bus1="bus_2",
    x=0.1,
    r=0.01,
    s_nom=200,
)

n.add(
    "Line",
    "line_2",
    bus0="bus_2",
    bus1="bus_3",
    x=0.1,
    r=0.01,
    s_nom=200,
)

n.add(
    "Line",
    "line_3",
    bus0="bus_3",
    bus1="bus_1",
    x=0.1,
    r=0.01,
    s_nom=150,
)
```

```python
n.optimize()
```
